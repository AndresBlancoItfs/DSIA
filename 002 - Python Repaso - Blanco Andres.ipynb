## Ejercicio 1 – Variables y operaciones

```
nombre = input("Ingrese su nombre: ")
edad = int(input("Ingrese su edad: "))
ciudad = input("Ingrese la ciudad de residencia: ")

print("Hola, mi nombre es",nombre,", tengo",edad, " años y vivo en",ciudad)

a=int(input("Ingrese el primer valor:"))
b=int(input("Ingrese el segundo valor:"))

print("La suma de",a,"y",b,"es igual a",a+b)
print("La resta de",a,"y",b,"es igual a",a-b)
print("La multiplicacion entre",a,"y",b,"es igual a",a*b)

if(b==0):
	print("No se puede dividir por cero")
else:
	print("La division entre",a,"y",b,"es igual a",a/b)
```

## Ejercicio 2 – Condicionales

```
edad = int(input("Ingrese su edad: "))

if(edad>17):
	print("Es mayor de edad")
else:
	print("Es menor de edad")
```

## Ejercicio 3 – Listas y bucles

```
lista=[]

for i in range(5):
	valor=int(input("Ingrese un numero: "))
	lista.append(valor)
suma=0
for i in range(5):
	print("El valor en la posicion",i,"es igual a",lista[i])
	suma=suma+lista[i]
print("La suma de todos los elementos es igual a",suma)
print("El promedio es igual a",suma/5)
```
